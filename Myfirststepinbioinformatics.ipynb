```python
# My first step in bioinformatics

sequence = "ATGCGTAGCTAG"
print("The length of the sequence is:", len(sequence))
```

```python
# Count each nucleotide in the DNA sequence
count_A = sequence.count("A")
count_T = sequence.count("T")
count_C = sequence.count("C")
count_G = sequence.count("G")

print("A:", count_A)
print("T:", count_T)
print("C:", count_C)
print("G:", count_G)
```

```python
total_length = len(sequence)
gc_content = ((count_G + count_C)/total_length)*100
print("%GC:", gc_content)
```

```python
def calculate_gc_content(sequence):
    """Calculate the GC content percentage of a DNA sequence."""
    count_G = sequence.count("G")
    count_C = sequence.count("C")
    total_length = len(sequence)
    if total_length == 0:
        return 0
    gc_content = ((count_G + count_C)/total_length)*100
    return gc_content
```

```python
gc = calculate_gc_content(sequence)
print("GC content (%):", gc)
```

```python
# compare DNA sequences
sequence1 = "ATGCGTAGCTAG"
sequence2 = "ATATATATAT"

gc1 = calculate_gc_content(sequence1)
gc2 = calculate_gc_content(sequence2)

print("GC% of sequence 1:", gc1)
print("GC% of sequence 2:", gc2)

if gc1 > gc2:
    print("Sequence 1 has more GC content.")
else:
    print("Sequence 2 has more GC content.")
```
